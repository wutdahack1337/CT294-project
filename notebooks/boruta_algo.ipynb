```python
import pandas as pd
from boruta import BorutaPy
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_csv("../datasets/energydata_complete.csv")
df["date"] = pd.to_datetime(df["date"])
df["Appliances"] = pd.to_numeric(df["Appliances"])
df.set_index("date", inplace=True)
df
```

```python
X = df.iloc[:,1:-2]
y = df.Appliances
X
```

```python
rf = RandomForestRegressor(n_jobs=-1, max_depth=7, random_state=1337)
feat_selector = BorutaPy(rf, n_estimators='auto', verbose=2, random_state=1337)
```

```python
feat_selector.fit(X, y)
```

```python
feat_selector.support_
```
